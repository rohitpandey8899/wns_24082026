# Attaching Custom Schema

```python
from pyspark.sql.types import *

USER_SCHEMA = StructType([
    StructField("id", IntegerType()),
    StructField("age", IntegerType()),
    StructField("gender", StringType()),
    StructField("designation", StringType()),
    StructField("salary", IntegerType())
])

df = spark.read.csv(
    path="/Volumes/wns24082026/quickstart_schema/sandbox/datasets/sample_dataset/users.csv",
    header=True,
    schema=USER_SCHEMA
)
display(df)
```

#Handling Bad record
PERMISSIVE.  # make a new column as corrupt record
dROPMALFORMED # just drop the corrupt record
FaILFAST # will not create a data frame

```python
spark.read.json(path="/Volumes/wns24082026/quickstart_schema/sandbox/datasets/access_logs.json",
                mode="PERMISSIVE").display()
```

```python
spark.read.json(path="/Volumes/wns24082026/quickstart_schema/sandbox/datasets/access_logs.json",
                mode="dROPMALFORMED").display()
```

```python
spark.read.json(path="/Volumes/wns24082026/quickstart_schema/sandbox/datasets/access_logs.json",
                mode="FaILFAST").display()
```

# Writing Mode

```python
df = spark.read.csv(
    path="/Volumes/wns24082026/quickstart_schema/sandbox/datasets/sample_dataset/users.csv",
    header=True,
    inferSchema=True,
)
df.write.save(
    format="PARQUET",
    path="/Volumes/wns24082026/quickstart_schema/sandbox/output/output_parquet",
)
 
```

```python
df = spark.read.csv(
    path="/Volumes/wns24082026/quickstart_schema/sandbox/datasets/sample_dataset/users.csv",
    header=True,
    inferSchema=True,
)
df.write.save(
    mode="OVERWRITE",
    format="PARQUET",
    path="/Volumes/wns24082026/quickstart_schema/sandbox/output/output_parquet"
)

spark.read.parquet("/Volumes/wns24082026/quickstart_schema/sandbox/output/output_parquet").count()
```

```python
df = spark.read.csv(
    path="/Volumes/wns24082026/quickstart_schema/sandbox/datasets/sample_dataset/users.csv",
    header=True,
    inferSchema=True,
)
df.write.save(
    mode="APPEND",
    format="PARQUET",
    path="/Volumes/wns24082026/quickstart_schema/sandbox/output/output_parquet"
)

spark.read.parquet("/Volumes/wns24082026/quickstart_schema/sandbox/output/output_parquet").count()
```
