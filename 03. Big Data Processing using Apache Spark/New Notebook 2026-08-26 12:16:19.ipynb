# Prerequisite

```python
df = spark.read.csv(path="/Volumes/wns24082026/quickstart_schema/sandbox/datasets/employee.csv",
                    header=True,
                    inferSchema=True,
                    sep="|",
                    quote="'")
display(df)
```

## select opration

```python
from pyspark.sql.functions import col
df.select(col("id"),col("name")).show();
```

```python
from pyspark.sql.functions import col
df.select(col("id").alias("user_id"),col("name")).show();
```

## transformation

```python
df.filter((col("desig")=="Developer") | (col("desig")=="Team Lead")).show()
```

```python
df.filter(col("desig").isin("Developer","Team Lead")).show()
```

```python
from pyspark.sql.functions import lower
df.filter(lower(col("desig")).like("%Dev%")).show()
```

#Derive Column

```python
from pyspark.sql.functions import lit
df.withColumn("catgory",lit("Not Define")).show()
```

```python
from pyspark.sql.functions import when

result_df = df.withColumn(
    "catgory",
    when(col("exp") >= 10,"Experienced")
    .when(col("exp") >= 5, "Senior")
    .when(col("exp") >= 0, "Mid")
    .otherwise("Invaild Exp"),
)

result_df.select("exp","catgory").show()
```

```python
result_df.groupBy("catgory").count().sort("catgory", accessding=True).show()
```

```python
result_df.groupBy("catgory").count().sort(col("catgory").desc()).show()
```

```python
from pyspark.sql.functions import min, max, avg
result_df.groupBy("catgory").agg(min("exp"),
                                  max("exp"),
                                  avg("exp")).show()
 
```
